```python
import os
import pickle
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
from videoutils import make_animation
```

```python
train_videos = pickle.load(open("train_structure.pkl", "rb"))
test_videos = pickle.load(open("test_structure.pkl", "rb"))
```

```python
train_videos.keys()
```

```python
def counter(dataset):
    frames_count = []

    for class_name, videos in dataset.items():
        for video_name, info in videos.items():
            row = {"vid_name": video_name}
            row["label"] = class_name
            row["num_frames"] = info["num_frames"]
            frames_count.append(row)
    
    return pd.DataFrame(frames_count)    
```

## Train dataset class statistics

```python
train_df = counter(train_videos)
train_df.to_csv("train_df.csv", sep=';')
train_df.describe()
```
