```python
from videoutils import make_animation
import pickle
import random
import numpy as np
import os
```

```python
train_videos = pickle.load(open("train_structure.pkl", "rb"))
test_videos = pickle.load(open("test_structure.pkl", "rb"))
```

```python
np.random.seed(0)
base_dir = "examples"
for class_name, videos in train_videos.items():
    filtered_videos = [
        v_name for v_name, v_data in videos.items() 
        if len(v_data["frames"]) <= 1000
    ]
    
    class_folder = os.path.join(base_dir, class_name)
    os.makedirs(class_folder, exist_ok=True)

    # 5/or less unique random videos
    selected_videos = np.random.choice(filtered_videos, size=min(5, len(filtered_videos)), replace=False)

    for video_name in selected_videos:
        frames_list = train_videos[class_name][video_name]["frames"]
        
        save_path = os.path.join(class_folder, f"{video_name}.mp4")
        make_animation(frames_list, save=True, save_path=save_path)
        
```
